```python
# Use-case: I want to create a new column called UpdatedYearlySalary based on the following increment/hike conditions
# If the salary is less than equal to 1500, then inc the sal by 10%
# If the salary is between 1501 and 10000, then inc the sal by 5%
# If the salary is between 10001 and 20000, then inc the sal by 2.5%
# If the salary is greater than 20000 , then no inc
```

```python
def yearlysalary(salary):
    if salary <= 1500:
        salary = salary * 1.10
    elif salary >= 1501 and salary <= 10000:
        salary = salary * 1.05
    elif salary >= 10001 and salary <= 2000:
        salary = salary * 1.025
    elif salary > 2000:
        salary = salary 
    return salary 
```

```python
df["yearlyupdatedsalary"] = df.esal.apply(yearlysalary)
```

```python
import pandas as pd

df = pd.DataFrame({
    "patient_id": [1, 2, 3, 4, 5, 6],
    "gender": ["M", "F", "F", "M", "F", "M"],
    "cancer_type": ["A", "A", "B", "B", "A", "B"],
    "survival_days": [200, 350, 400, 150, 500, 300]
})

df
```

```python
df.groupby("cancer_type")["survival_days"].mean()
```

```python
df.groupby("cancer_type")["survival_days"].agg(["mean", "median", "count"])
```

```python
df.groupby(["cancer_type", "gender"])["survival_days"].mean()
```

```python
df.groupby("cancer_type")["survival_days"].mean().reset_index()
```
